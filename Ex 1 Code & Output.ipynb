```python
import ipywidgets as widgets
from IPython.display import display

slider = widgets.IntSlider(
    value=50,
    min=0,
    max=100,
    step=1,
    description='Value:'
)

output = widgets.Output()

def ab(change):
    with output:
        output.clear_output()
        print(f'Slider value is {change["new"]}')

slider.observe(ab, names='value')

display(slider, output)
```
